```python
# import necessary libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Your task is to produce visualizations in Python that show:
For each city:
 - The average house sale price
 - The average estimated monthly rental income
 - How long it will take a property on average to achieve the sale price from rental income for a selected city.

The visualizations should adhere to User Experience best practices including:

- Inclusive & accessible design
- Appropriate use of visualizations

```python
# import data

ss_data = pd.read_csv('Resources/Data/san_berandino_prices.csv')

display(ss_data.head())
```

```python
q1, q2, q3 = ss_data['Price'].quantile([0.25,0.5,0.75])
```

```python
IRQ = (q3 - q1)
```

```python
fil = ss_data[ss_data['Price']< q2 + 1.5 * IRQ]
```

```python
sns.boxplot(x='City', y='Price', data=ss_data, showmeans=True)
```

```python
sns.boxplot(x='City', y='Price', data=fil)
```

```python
sns.histplot(data=ss_data, x='Price', hue='City', bins=30)
```

```python
g = sns.FacetGrid(ss_data, col='City')
g.map_dataframe(sns.histplot, x='Price')
```

```python
display(ss_data.groupby('City')['Price'].mean())
```

```python
ss_data.groupby('City')['Price'].median()
```

```python
df = ss_data[ss_data['Price'] < 3000000]
```

```python
display(df.groupby('City')['Price'].mean())
```

```python
df.groupby('City')['Price'].median()
```

```python
g = sns.FacetGrid(df, col='City')
g.map_dataframe(sns.histplot, x='Price')
```
